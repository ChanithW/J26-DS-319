```python
import json, tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.metrics import average_precision_score, precision_recall_fscore_support

OUT = '/content/drive/MyDrive/Component1/dataset'
MOD = '/content/drive/MyDrive/Component1/model'
os.makedirs(MOD, exist_ok=True)
SEED = 42
keras.utils.set_random_seed(SEED)                     # same results every run

# ---- 1. load model inputs ----
d = np.load(f'{OUT}/model_inputs.npz', allow_pickle=True)
X, S, yF, yE, hasE, split = d['X'], d['S'], d['y_flare'], d['y_E'], d['has_E'], d['split']
yE0 = np.nan_to_num(yE)                               # NaN → 0 (masked out by the weight below)
wE  = hasE.astype(np.float32)                         # energy weight: 1 = has energy, 0 = skip

def part(name):
    m = split == name
    return ({'light_curve': X[m], 'star': S[m]},
            {'flare': yF[m], 'energy': yE0[m]},
            {'flare': np.ones(m.sum(), np.float32), 'energy': wE[m]})

tr, va = part('train'), part('val')

# ---- 2. build the model ----
def conv_block(x, filters, kernel, pool=True):
    x = layers.Conv1D(filters, kernel, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    return layers.MaxPooling1D(2)(x) if pool else x

def pooled(x):
    return layers.Dropout(0.3)(layers.Concatenate()([layers.GlobalAveragePooling1D()(x),
                                                      layers.GlobalMaxPooling1D()(x)]))

lc_in   = keras.Input((128, 2), name='light_curve')
star_in = keras.Input((2,), name='star')

x = conv_block(lc_in, 32, 7)                          # shared block 1: simple slopes
x = conv_block(x, 64, 5)                              # shared block 2: rise-and-fall patterns
x = conv_block(x, 128, 5)                             # shared block 3: whole flare shape

f_det = pooled(conv_block(x, 128, 3, pool=False))     # block 4, detection only
f_en  = pooled(conv_block(x, 128, 3, pool=False))     # block 4, energy only

det = layers.Dense(64, activation='relu')(f_det)
det_out = layers.Dense(1, activation='sigmoid', name='flare')(det)          # flare probability

s = layers.Dense(16, activation='relu')(star_in)
e = layers.Concatenate()([f_en, s])                                          # flare features + star
e = layers.Dropout(0.3)(layers.Dense(64, activation='relu')(e))
en_out = layers.Dense(1, name='energy')(e)                                   # scaled log10 energy

model = keras.Model({'light_curve': lc_in, 'star': star_in}, {'flare': det_out, 'energy': en_out})
model.compile(optimizer=keras.optimizers.Adam(1e-3),
              loss={'flare': 'binary_crossentropy', 'energy': keras.losses.Huber()},
              loss_weights={'flare': 1.0, 'energy': 2.0})
model.summary()

# ---- 3. train with early stopping ----
cb = [keras.callbacks.EarlyStopping(monitor='val_loss', patience=6, restore_best_weights=True),
      keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3)]
hist = model.fit(tr[0], tr[1], sample_weight=tr[2],
                 validation_data=va, epochs=60, batch_size=128, callbacks=cb, verbose=2)

# ---- 4. choose the detection threshold on VALIDATION (best F1) ----
pv = model.predict(va[0], batch_size=1024, verbose=0)
p_flare = pv['flare'][:, 0]
best = max(((t, precision_recall_fscore_support(va[1]['flare'], p_flare >= t, average='binary',
                                                zero_division=0)[2]) for t in np.arange(0.05, 0.96, 0.01)),
           key=lambda z: z[1])
THR = float(round(best[0], 2))

# ---- 5. validation results ----
prec, rec, f1, _ = precision_recall_fscore_support(va[1]['flare'], p_flare >= THR, average='binary')
print(f'\nThreshold {THR} | precision {prec:.3f} | recall {rec:.3f} | F1 {f1:.3f} | '
      f'AP {average_precision_score(va[1]["flare"], p_flare):.3f}')

st = json.load(open(f'{OUT}/prep_settings.json'))
m = va[2]['energy'] > 0
pred_logE = pv['energy'][m, 0] * st['logE_std'] + st['logE_mean']
true_logE = va[1]['energy'][m] * st['logE_std'] + st['logE_mean']
err = np.abs(pred_logE - true_logE)
bins = [-np.inf, 32, 33, 34, np.inf]
type_acc = (np.digitize(pred_logE, bins[1:-1]) == np.digitize(true_logE, bins[1:-1])).mean()
print(f'Energy: median error {np.median(err):.2f} dex | mean error {err.mean():.2f} dex | '
      f'within a factor of 2: {(err < 0.301).mean():.0%} | type correct: {type_acc:.0%}')

# ---- 6. save ----
model.save(f'{MOD}/best.keras')
json.dump({'threshold': THR, 'seed': SEED, 'epochs_trained': len(hist.history['loss']),
           'energy_loss_weight': 2.0}, open(f'{MOD}/settings.json', 'w'), indent=2)
print('Saved model and settings to', MOD)

# ---- 7. learning curves ----
plt.figure(figsize=(6, 3))
plt.plot(hist.history['loss'], label='train'); plt.plot(hist.history['val_loss'], label='validation')
plt.xlabel('epoch'); plt.ylabel('total loss'); plt.legend(); plt.show()
```

```python
import json, tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
from sklearn.metrics import average_precision_score, precision_recall_fscore_support

OUT = '/content/drive/MyDrive/Component1/dataset'
MOD = '/content/drive/MyDrive/Component1/model'
SEED, DROPOUT, L2, LR, MAX_SHIFT, NOISE = 42, 0.5, 1e-4, 5e-4, 8, 0.2
keras.utils.set_random_seed(SEED)

# ---- 1. load model inputs ----
d = np.load(f'{OUT}/model_inputs.npz', allow_pickle=True)
X, S, yF, yE, hasE, split = d['X'], d['S'], d['y_flare'], d['y_E'], d['has_E'], d['split']
yE0, wE = np.nan_to_num(yE), hasE.astype(np.float32)

def part(name):
    m = split == name
    return ({'light_curve': X[m], 'star': S[m]},
            {'flare': yF[m], 'energy': yE0[m]},
            {'flare': np.ones(m.sum(), np.float32), 'energy': wE[m]})

tr, va = part('train'), part('val')

# ---- 2. data augmentation (training only) ----
def augment(inputs, targets, weights):
    lc = inputs['light_curve']
    shift = tf.random.uniform([], -MAX_SHIFT, MAX_SHIFT + 1, dtype=tf.int32)
    lc = tf.roll(lc, shift=shift, axis=0)                                   # move the window by a few points
    noise = tf.random.normal(tf.shape(lc[:, :1]), stddev=NOISE)
    lc = tf.concat([lc[:, :1] + noise, lc[:, 1:]], axis=1)                  # a little extra noise on channel 1
    return {'light_curve': lc, 'star': inputs['star']}, targets, weights

train_ds = (tf.data.Dataset.from_tensor_slices(tr)
            .shuffle(len(tr[0]['star']), seed=SEED)
            .map(augment, num_parallel_calls=tf.data.AUTOTUNE)
            .batch(128).prefetch(tf.data.AUTOTUNE))
val_ds = tf.data.Dataset.from_tensor_slices(va).batch(1024)

# ---- 3. model with more dropout + L2 ----
reg = regularizers.l2(L2)

def conv_block(x, filters, kernel, pool=True):
    x = layers.Conv1D(filters, kernel, padding='same', use_bias=False, kernel_regularizer=reg)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    return layers.MaxPooling1D(2)(x) if pool else x

def pooled(x):
    return layers.Dropout(DROPOUT)(layers.Concatenate()([layers.GlobalAveragePooling1D()(x),
                                                          layers.GlobalMaxPooling1D()(x)]))

lc_in, star_in = keras.Input((128, 2), name='light_curve'), keras.Input((2,), name='star')
x = conv_block(lc_in, 32, 7); x = conv_block(x, 64, 5); x = conv_block(x, 128, 5)      # shared
f_det = pooled(conv_block(x, 128, 3, pool=False))                                        # detection block 4
f_en  = pooled(conv_block(x, 128, 3, pool=False))                                        # energy block 4

det = layers.Dropout(DROPOUT)(layers.Dense(64, activation='relu', kernel_regularizer=reg)(f_det))
det_out = layers.Dense(1, activation='sigmoid', name='flare')(det)

s = layers.Dense(16, activation='relu', kernel_regularizer=reg)(star_in)
e = layers.Concatenate()([f_en, s])
e = layers.Dropout(DROPOUT)(layers.Dense(64, activation='relu', kernel_regularizer=reg)(e))
en_out = layers.Dense(1, name='energy')(e)

model = keras.Model({'light_curve': lc_in, 'star': star_in}, {'flare': det_out, 'energy': en_out})
model.compile(optimizer=keras.optimizers.Adam(LR),
              loss={'flare': 'binary_crossentropy', 'energy': keras.losses.Huber()},
              loss_weights={'flare': 1.0, 'energy': 2.0})

# ---- 4. train: checkpoint the best, more patience ----
cb = [keras.callbacks.ModelCheckpoint(f'{MOD}/best_v2.keras', monitor='val_loss', save_best_only=True),
      keras.callbacks.EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True),
      keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=4)]
hist = model.fit(train_ds, validation_data=val_ds, epochs=80, callbacks=cb, verbose=2)

# ---- 5. threshold on VALIDATION + results ----
pv = model.predict(va[0], batch_size=1024, verbose=0)
p = pv['flare'][:, 0]
THR = float(round(max(np.arange(0.05, 0.96, 0.01),
                      key=lambda t: precision_recall_fscore_support(va[1]['flare'], p >= t, average='binary',
                                                                    zero_division=0)[2]), 2))
prec, rec, f1, _ = precision_recall_fscore_support(va[1]['flare'], p >= THR, average='binary')
print(f'\nv2 | threshold {THR} | precision {prec:.3f} | recall {rec:.3f} | F1 {f1:.3f} | '
      f'AP {average_precision_score(va[1]["flare"], p):.3f}')

st = json.load(open(f'{OUT}/prep_settings.json'))
m = va[2]['energy'] > 0
pred = pv['energy'][m, 0] * st['logE_std'] + st['logE_mean']
true = va[1]['energy'][m] * st['logE_std'] + st['logE_mean']
err = np.abs(pred - true)
cuts = [32, 33, 34]
print(f'v2 | energy median error {np.median(err):.2f} dex | mean {err.mean():.2f} dex | '
      f'within x2: {(err < 0.301).mean():.0%} | type correct: {(np.digitize(pred, cuts) == np.digitize(true, cuts)).mean():.0%}')
print('v1 | threshold 0.46 | precision 0.851 | recall 0.746 | F1 0.795 | AP 0.878')
print('v1 | energy median error 0.18 dex | mean 0.24 dex | within x2: 71% | type correct: 77%')

# ---- 6. save settings ----
json.dump({'version': 'v2', 'threshold': THR, 'seed': SEED, 'dropout': DROPOUT, 'l2': L2, 'lr': LR,
           'augmentation': {'max_shift': MAX_SHIFT, 'noise_ch1': NOISE},
           'epochs_trained': len(hist.history['loss']), 'energy_loss_weight': 2.0},
          open(f'{MOD}/settings_v2.json', 'w'), indent=2)

# ---- 7. learning curves (total, detection, energy) ----
fig, ax = plt.subplots(1, 3, figsize=(15, 3))
for a, k, t in zip(ax, ['loss', 'flare_loss', 'energy_loss'], ['total', 'detection', 'energy']):
    a.plot(hist.history[k], label='train'); a.plot(hist.history['val_' + k], label='validation')
    a.set_title(f'{t} loss'); a.set_xlabel('epoch'); a.legend()
plt.tight_layout(); plt.show()
```

```python
import json, tensorflow as tf
from tensorflow import keras
from sklearn.metrics import (average_precision_score, precision_recall_fscore_support,
                             confusion_matrix, r2_score)

OUT = '/content/drive/MyDrive/Component1/dataset'
MOD = '/content/drive/MyDrive/Component1/model'
RES = '/content/drive/MyDrive/Component1/results'
os.makedirs(RES, exist_ok=True)

# ---- 1. load inputs, star types and SNR ----
d = np.load(f'{OUT}/model_inputs.npz', allow_pickle=True)
X, S, yF, yE, hasE, split = d['X'], d['S'], d['y_flare'], d['y_E'], d['has_E'], d['split']
star = np.load(f'{OUT}/flare_dataset_split.npz', allow_pickle=True)['star']
teff = star[:, 0].astype(float)
spt = np.select([teff < 3900, teff < 5300, teff < 6000], ['M', 'K', 'G'], 'F')   # star type from Teff
snr = X[:, :, 0].max(axis=1)          # peak of channel 1 = peak (flux-1)/flux_err
low = snr < 5                          # low-SNR flares

# ---- 2. load the v2 model and predict every window ----
model = keras.models.load_model(f'{MOD}/best_v2.keras')
THR = json.load(open(f'{MOD}/settings_v2.json'))['threshold']
st = json.load(open(f'{OUT}/prep_settings.json'))
pred = model.predict({'light_curve': X, 'star': S}, batch_size=1024, verbose=0)
p = pred['flare'][:, 0]
is_flare = p >= THR
logE_pred = pred['energy'][:, 0] * st['logE_std'] + st['logE_mean']     # back to log10 erg
logE_true = yE * st['logE_std'] + st['logE_mean']                       # NaN where no energy

TYPES, CUTS = ['Solar-like', 'Large', 'Superflare', 'Extreme'], [32, 33, 34]

# ---- 3. one function that measures everything for a group of windows ----
def evaluate(m, set_name, group='all'):
    r = {'set': set_name, 'group': group}
    t, f = yF[m] == 1, is_flare[m]
    r['n_flares'], r['n_quiet'] = int(t.sum()), int((~t).sum())
    r['precision'], r['recall'], r['f1'], _ = precision_recall_fscore_support(
        t, f, average='binary', zero_division=0)
    r['ap'] = average_precision_score(t, p[m]) if 0 < t.sum() < len(t) else np.nan
    lf = m & (yF == 1) & low
    r['n_low_snr'] = int(lf.sum())
    r['low_snr_recall'] = is_flare[lf].mean() if lf.any() else np.nan
    e = m & hasE                                                       # energy: all true flares
    err = np.abs(logE_pred[e] - logE_true[e])
    r['median_err'] = np.median(err) if e.any() else np.nan
    r['mean_err'] = err.mean() if e.any() else np.nan
    r['r2'] = r2_score(logE_true[e], logE_pred[e]) if e.sum() > 1 else np.nan
    r['within_x2'] = (err < 0.301).mean() if e.any() else np.nan
    r['type_acc'] = (np.digitize(logE_pred[e], CUTS) == np.digitize(logE_true[e], CUTS)).mean() if e.any() else np.nan
    el = e & low
    r['low_snr_median_err'] = np.median(np.abs(logE_pred[el] - logE_true[el])) if el.any() else np.nan
    return r

rows = []
for s in ['val', 'test_lin', 'test_gunther']:
    m = split == s
    rows.append(evaluate(m, s))
    for t in 'FGKM':
        rows.append(evaluate(m & (spt == t), s, t))
res = pd.DataFrame(rows).round(3)
res.to_csv(f'{RES}/test_results_v2.csv', index=False)

pd.set_option('display.width', 200)
print(f'Threshold {THR} (chosen on validation)\n')
print('DETECTION')
print(res[['set', 'group', 'n_flares', 'n_quiet', 'precision', 'recall', 'f1', 'ap',
           'n_low_snr', 'low_snr_recall']].to_string(index=False))
print('\nENERGY (median/mean error in dex)')
print(res[['set', 'group', 'median_err', 'mean_err', 'r2', 'within_x2', 'type_acc',
           'low_snr_median_err']].to_string(index=False))

# ---- 4. confusion matrices ----
for s in ['test_lin', 'test_gunther']:
    m = split == s
    cm = confusion_matrix(yF[m], is_flare[m])
    print(f'\n{s}: detection confusion matrix')
    print(pd.DataFrame(cm, index=['true quiet', 'true flare'], columns=['pred quiet', 'pred flare']))
    e = m & hasE
    ct = pd.crosstab(pd.Categorical(np.array(TYPES)[np.digitize(logE_true[e], CUTS)], TYPES),
                     pd.Categorical(np.array(TYPES)[np.digitize(logE_pred[e], CUTS)], TYPES),
                     rownames=['true type'], colnames=['predicted type'], dropna=False)
    print(f'\n{s}: energy type confusion matrix')
    print(ct)

# ---- 5. predicted vs true energy ----
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
for a, s in zip(ax, ['test_lin', 'test_gunther']):
    e = (split == s) & hasE
    a.scatter(logE_true[e & ~low], logE_pred[e & ~low], s=3, alpha=0.3, label='SNR ≥ 5')
    a.scatter(logE_true[e & low], logE_pred[e & low], s=3, alpha=0.3, label='low SNR (< 5)')
    lim = [30, 37]
    a.plot(lim, lim, 'k-', lw=1)
    a.plot(lim, [lim[0] + 0.301, lim[1] + 0.301], 'k--', lw=0.7)
    a.plot(lim, [lim[0] - 0.301, lim[1] - 0.301], 'k--', lw=0.7)
    for c in CUTS:
        a.axvline(c, color='grey', lw=0.5); a.axhline(c, color='grey', lw=0.5)
    a.set_xlim(lim); a.set_ylim(lim)
    a.set_xlabel('catalogue log10 E (erg)'); a.set_ylabel('predicted log10 E (erg)')
    a.set_title(s); a.legend(markerscale=4)
plt.tight_layout(); plt.savefig(f'{RES}/energy_scatter_v2.png', dpi=150); plt.show()
```
